## ***indexing***  [start : end : step]

```python
my_phone_number = '01012345678'
print(my_phone_number[0:3]) # it will print the first 3 digits
print(my_phone_number[3:6]) # it will print the next 3 digits
print(my_phone_number[6:]) # it will print the last 5 digits
print(my_phone_number[-4:]) # it will print the last 4 digits
print(my_phone_number[::2]) # it will print every second digit
print(my_phone_number[::-1]) # it will print the string in reverse order
print(my_phone_number[1:8:2]) # it will print every second digit from index 1 to 7
```

## ***format specifiers***  {value : flags}

```python
price1= 3000.14159
price2= -987.65
price3= 12.34

print(f"price1 is ${price1:+,.1f}")
print(f"price2 is ${price2:10}")
print(f"price3 is ${price3:>11}")
```
