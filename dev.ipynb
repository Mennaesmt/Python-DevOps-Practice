## ***Server Health Monitoring***


```python
Disk_usage = float(input("enter the Disk usage :"))

if  Disk_usage < 70:
    print("server is healthy.")
elif Disk_usage >= 70 and Disk_usage < 90:
    print("server is warning.")
else:
    print("server is in danger.")
    
```
